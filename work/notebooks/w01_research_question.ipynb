## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

**Freestyle.**

**Research question:** Can historical search and engagement signals identify webpages that appear stable today but are at risk of future performance decline?

>I want to explore whether search and engagement signals can identify pages that appear healthy today but are at risk of declining in the near future.
Rather than reacting after performance has already dropped, I want to investigate whether historical patterns can support earlier, evidence-based review decisions.
I chose this direction because helping editors prioritize pages before visible decline has a more practical impact than simply explaining why pages have already lost performance.


## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

>This work aims to improve the decision of which pages should be reviewed before their performance visibly declines.
The primary users are content managers or editors who have limited time and cannot review every page.
A correct recommendation allows them to intervene earlier, while an incorrect recommendation either wastes editorial effort on healthy pages or misses pages that continue to lose visibility without review.
