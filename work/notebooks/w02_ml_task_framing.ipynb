## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Classification.**

>My project will use **binary classification** to predict whether a webpage is likely to experience future visibility decline based on historical search and engagement signals.
The model's predicted probabilities will then be used to rank pages by risk, helping content managers prioritize which pages should be reviewed first.
I chose this approach because the statistical problem will be a binary prediction, while the real decision will be deciding which pages deserve attention first given limited editorial resources. A fixed probability threshold, such as 0.5, will therefore not automatically determine which pages should be reviewed.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

>The target is a **later observed visibility decline**, measured after the historical feature window. For the initial development specification, a page is positive when its impressions in the later 30-day window are more than 20% below its impressions in the preceding 30-day historical window:
>
>`is_declining = 1 if later_impressions < 0.80 × prior_impressions`
>
>This is an operational definition rather than a universal definition of content decay. The target must come from a later observed outcome, not from an existing FlyRank product flag or from a field that is also used as a model input. A minimum historical-impression requirement will later be used to reduce unstable percentage changes from very small denominators. If the project later adds a longer historical window or an explicit stable-at-decision population, that will be treated as a revised modeling specification and validated again.


## 3. Success metric

*One metric you can defend. What number means 'good'?*

>My primary operational metrics are **Precision@20 and Precision@50** because the final output is a ranked review queue and a content team can only inspect a limited number of pages. These metrics answer: of the first 20 or 50 pages sent for review, how many later meet the defined decline outcome? I will also report **Average Precision** for the overall ranking, **ROC-AUC** as a supporting discrimination metric, and the **decline base rate** so that Precision@K has proper context. Later reporting can also show lift@K relative to that base rate.


## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

>The unit of analysis is one content page.
Each row represents a single pseudonymized webpage together with its historical search, engagement, and content-related measurements.
The model will use these page-level observations to learn patterns associated with future performance decline.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

>FlyRank already uses rule-based thresholds to flag content opportunities, catching conditions once they become clear and visible.
However, fixed rules can only respond after those conditions have already been met.
My project explores whether historical search and engagement signals can be combined to identify pages that may be at risk earlier.
Since page performance depends on several interacting signals rather than a single condition, machine learning is better suited to learn these patterns than relying on fixed thresholds alone.
