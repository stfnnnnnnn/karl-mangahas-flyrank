## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

>My project is primarily a **classification task that will be used for ranking**. I want to estimate whether a page is at risk of a later decline, which gives me a yes/no outcome for model training. However, the practical decision from Week 1 is which pages a content manager should inspect first, so I expect to use the model's score or probability to rank pages rather than treating the classification threshold itself as the final action.
>
>At this stage I am not choosing a specific algorithm yet. I first want to define the outcome and evaluation clearly enough that I can later compare a simple rule with learned models.


## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

>My intended target is a **future decline in search visibility**, most likely defined from a later change in search impressions. The basic shape I want is:
>
>```text
>historical measurements → decision point → later decline outcome
>```
>
>I have not finalized the exact window lengths or decline threshold yet. Those choices depend on what the warehouse history supports and will be specified in the data-contract stage.
>
>The starter dataset already contains a decline label based on `trend_direction`, but that is calculated from the starter snapshot's own comparison windows. I therefore treat it as a useful **proxy/example label for learning the workflow**, not automatically as the final capstone target. I would prefer the warehouse project target to come from an observed later period rather than from an existing product decision or current-window flag.


## 3. Success metric

*One metric you can defend. What number means 'good'?*

>Because the eventual output should help decide **which pages to review first**, my main operational metric will be **Precision@K**. For example, Precision@20 or Precision@50 would tell me how many pages in the first 20 or 50 recommendations actually meet the decline outcome.
>
>I also plan to report **Average Precision** to evaluate the ranking more broadly and **ROC-AUC** as a supporting classification metric. I should always compare these results with the underlying decline rate so that a seemingly good Precision@K number has context.
>
>Before relying on machine learning, I need a **transparent rule-based baseline**. The learned model should only be considered useful if it improves the ranking enough to justify the extra complexity. I am not fixing the exact baseline formula yet; that belongs in the baseline-building week.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

>A fixed rule may already be useful, so I do not assume that ML will automatically be better. A simple rule might prioritize pages using one or two conditions such as meaningful visibility and signs of weakness.
>
>ML may become useful if several historical signals interact in ways that are difficult to express with one stable threshold—for example, visibility level, clicks, position, recent movement, content age/freshness, or engagement context may matter differently in combination. A learned model could potentially combine those signals into a better ranking.
>
>The comparison is therefore part of the research question: **does a learned ranking actually improve on a transparent rule on unseen data?** If it does not, the simpler rule is the better answer.
