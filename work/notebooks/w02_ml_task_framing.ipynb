## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
Classification.

My project aims to classify whether a webpage is likely to experience future performance decline based on historical search and engagement signals.
The model's predicted probabilities will then be used to rank pages by risk, helping content managers prioritize which pages should be reviewed first.
I chose this approach because the underlying problem is a binary prediction, while the real decision is deciding which pages deserve attention first given limited editorial resources.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
The target will be an observed future decline in page performance, measured using a later time window instead of the same period used to create the features.
During the early stages of the internship, the starter dataset's trend_direction serves only as a workflow proxy.
The final project will instead use an observed future outcome from the warehouse to avoid data leakage.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
Since the model's predictions will be used to prioritize which pages should be reviewed first, Precision@K is the primary success metric.
A good result means that most of the highest-ranked pages identified by the model genuinely require attention, allowing limited editorial resources to focus on the pages most likely to benefit from early review.
While the underlying model performs binary classification, its predictions are ultimately used to rank pages for decision-making.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
The unit of analysis is one content page.
Each row represents a single pseudonymized webpage together with its historical search, engagement, and content-related measurements.
The model will use these page-level observations to learn patterns associated with future performance decline.
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
FlyRank already uses rule-based thresholds to flag content opportunities, catching conditions once they become clear and visible.
However, fixed rules can only respond after those conditions have already been met.
My project explores whether historical search and engagement signals can be combined to identify pages that may be at risk earlier.
Since page performance depends on several interacting signals rather than a single condition, machine learning is better suited to learn these patterns than relying on fixed thresholds alone.
```
